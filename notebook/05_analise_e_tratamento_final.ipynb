## TRATAMENTO FINAL: COLUNAS DE MÚLTIPLA ESCOLHA (AGORA COM SEGURANÇA)

```python
print("\nProcessando 'linguagens_usadas_dia_a_dia' com get_dummies...")
df_final['linguagens_usadas_dia_a_dia'].fillna('Nenhuma', inplace=True)
linguagens_dummies = df_final['linguagens_usadas_dia_a_dia'].str.get_dummies(sep=', ')
linguagens_dummies.columns = [f'usa_linguagem_{col.strip().lower().replace(" ", "_").replace("/", "_").replace(".", "")}' for col in linguagens_dummies.columns]
df_final = pd.concat([df_final, linguagens_dummies], axis=1)
print("Colunas de linguagens recriadas e adicionadas!")
```

```python
if 'ferramentas_bi_usadas_dia_a_dia' in df_final.columns:
  print("\nProcessando 'ferramentas_bi_usadas_dia_a_dia' com get_dummies...")
  df_final['ferramentas_bi_usadas_dia_a_dia'].fillna('Nenhuma', inplace=True)
  bi_dummies = df_final['ferramentas_bi_usadas_dia_a_dia'].str.get_dummies(sep=', ')
  bi_dummies.columns = [f'usa_bi_{col.strip().lower().replace(" ", "_").replace("/", "_").replace("(", "").replace(")", "")}' for col in bi_dummies.columns]
  df_final = pd.concat([df_final, bi_dummies], axis=1)
  print("Colunas de ferramentas de BI recriadas e adicionadas!")
```

```python
# ==============================================================================
# CONSOLIDAÇÃO SIMPLIFICADA DAS COLUNAS DE FERRAMENTAS DE BI (VERSÃO RECOMENDADA)
# ==============================================================================
print("\n--- Consolidando e Agrupando Ferramentas de BI (Versão Simplificada) ---")

# --- 1. Consolidação de Nomes Variados (Como antes) ---
# Google Data Studio / Looker Studio
cols_looker = ['usa_bi_google_data_studio', 'usa_bi_looker_studio_antigo_google_data_studio', 'usa_bi_looker','usa_bi_looker_studio']
df_final['usa_bi_looker_studio_consolidado'] = df_final[[col for col in cols_looker if col in df_final.columns]].sum(axis=1).clip(upper=1)
df_final.drop(columns=[col for col in cols_looker if col in df_final.columns], inplace=True, errors='ignore')



# Nenhuma Ferramenta
cols_nenhuma = ['usa_bi_nenhum', 'usa_bi_nenhuma', 'usa_bi_nenhuma.1', 'usa_bi_não_utilizo', 
                'usa_bi_não_utilizo_ferramentas_de_bi', 'usa_bi_não_utilizo_nenhuma_ferramenta_de_bi_no_trabalho', 
                'usa_bi_não_utilizo_ferramentas_de_bi_ainda', 'usa_bi_nao_atuo_na_area_de_dados', # Adicionando relacionados
                'usa_bi_nāo_estou_trabalhando_com_ti_no_momento', 'usa_bi_estudante']
df_final['usa_bi_nenhuma_consolidado'] = df_final[[col for col in cols_nenhuma if col in df_final.columns]].sum(axis=1).clip(upper=1)
df_final.drop(columns=[col for col in cols_nenhuma if col in df_final.columns], inplace=True, errors='ignore')



# Amazon Quicksight
cols_quicksight = ['usa_bi_amazon_quicksight', 'usa_bi_amazon_quicksight.1']
df_final['usa_bi_amazon_quicksight_consolidado'] = df_final[[col for col in cols_quicksight if col in df_final.columns]].sum(axis=1).clip(upper=1)
df_final.drop(columns=[col for col in cols_quicksight if col in df_final.columns], inplace=True, errors='ignore')



# Streamlit
cols_streamlit = ['usa_bi_streamlit', 'usa_bi_streamlit.1']
df_final['usa_bi_streamlit_consolidado'] = df_final[[col for col in cols_streamlit if col in df_final.columns]].sum(axis=1).clip(upper=1)
df_final.drop(columns=[col for col in cols_streamlit if col in df_final.columns], inplace=True, errors='ignore')



# --- 2. Agrupamento de Categorias Lógicas (Como antes) ---
# Ferramentas Proprietárias
cols_propria = ['usa_bi_aplicação_própria_da_empresa_para_bi_dashboards', 'usa_bi_ferramenta_própria', 'usa_bi_solução_da_própria_empresa']
df_final['usa_bi_ferramenta_propria'] = df_final[[col for col in cols_propria if col in df_final.columns]].sum(axis=1).clip(upper=1)
df_final.drop(columns=[col for col in cols_propria if col in df_final.columns], inplace=True, errors='ignore')



# Excel / Google Sheets
cols_sheets = ['usa_bi_excel', 'usa_bi_fazemos_todas_as_análises_utilizando_apenas_excel_ou_planilhas_do_google']
df_final['usa_bi_excel_gsheets_consolidado'] = df_final[[col for col in cols_sheets if col in df_final.columns]].sum(axis=1).clip(upper=1)
df_final.drop(columns=[col for col in cols_sheets if col in df_final.columns], inplace=True, errors='ignore')



# BI Baseado em Código (Python/R) - Mantendo a ideia
cols_codigo = ['usa_bi_jupyter_notebook', 'usa_bi_pacotes_python', 'usa_bi_pandas', 
            'usa_bi_python_seaborn_matplotlib', 'usa_bi_rstudio', 'usa_bi_rmarkdown', 
            'usa_bi_streamlit_consolidado', 'usa_bi_shiny', 'usa_bi_shiny_apps', 'usa_bi_plotly'] # Usando a consolidada
df_final['usa_bi_codigo_python_r'] = df_final[[col for col in cols_codigo if col in df_final.columns]].sum(axis=1).clip(upper=1)
df_final.drop(columns=[col for col in cols_codigo if col in df_final.columns], inplace=True, errors='ignore')



# --- 3. Agrupamento Final em "Outras Ferramentas" ---
# Lista das ferramentas PRINCIPAIS que queremos MANTER individuais
ferramentas_principais = [
    'usa_bi_microsoft_powerbi', 
    'usa_bi_tableau',
    'usa_bi_qlik_view_qlik_sense',
    'usa_bi_looker_studio_consolidado', # Já consolidada
    'usa_bi_metabase',
    'usa_bi_superset',
    'usa_bi_amazon_quicksight_consolidado', # Já consolidada
    'usa_bi_sap_business_objects_sap_analytics',
    'usa_bi_alteryx' # Alteryx é mais ETL/Analytics, mas relevante
]

# Lista das categorias AGRUPADAS que queremos MANTER
categorias_agrupadas = [
    'usa_bi_nenhuma_consolidado',
    'usa_bi_excel_gsheets_consolidado',
    'usa_bi_ferramenta_propria',
    'usa_bi_codigo_python_r'
]

# Identifica TODAS as colunas de BI restantes que NÃO estão nas listas acima
cols_outras = [col for col in df_final.columns 
            if col.startswith('usa_bi_') 
            and col not in ferramentas_principais
            and col not in categorias_agrupadas]

print(f"\nAgrupando {len(cols_outras)} ferramentas menos comuns em 'usa_bi_outras'...")

# Cria a coluna 'Outras' somando todas as colunas identificadas
df_final['usa_bi_outras'] = df_final[cols_outras].sum(axis=1).clip(upper=1)

# Remove as colunas originais que foram agrupadas em 'Outras'
df_final.drop(columns=cols_outras, inplace=True, errors='ignore')

# --- 4. Renomeação Final (Simplificação) ---
df_final.rename(columns={
    'usa_bi_microsoft_powerbi': 'usa_bi_powerbi',
    'usa_bi_qlik_view_qlik_sense': 'usa_bi_qlik',
    'usa_bi_sap_business_objects_sap_analytics': 'usa_bi_sap',
    'usa_bi_looker_studio_consolidado': 'usa_bi_looker_studio',
    'usa_bi_nenhuma_consolidado': 'usa_bi_nenhuma',
    'usa_bi_amazon_quicksight_consolidado': 'usa_bi_amazon_quicksight',
    'usa_bi_excel_gsheets_consolidado': 'usa_bi_excel_gsheets'
}, inplace=True)

# --- 5. Verificação Final ---
colunas_bi_final = sorted([col for col in df_final.columns if col.startswith('usa_bi_')])
print(f"\n📊 RESULTADO DA CONSOLIDAÇÃO SIMPLIFICADA:")
print(f"Número final de colunas de BI: {len(colunas_bi_final)}")
print("Colunas de BI Finais:")
print(colunas_bi_final)

# Estatísticas de uso para validação
print(f"\n📈 ESTATÍSTICAS DE USO FINAIS:")
for col in colunas_bi_final:
    count = df_final[col].sum()
    percent = (count / len(df_final)) * 100
    print(f"  {col}: {count} usuários ({percent:.1f}%)")
```
