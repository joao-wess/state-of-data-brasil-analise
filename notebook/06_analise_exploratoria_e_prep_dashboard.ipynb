## 2 - TABELA DE POPULARIDADE DE LINGUAGENS USADAS POR ANO

```python
colunas_linguagens = [col for col in df_final.columns if col.startswith('usa_linguagem_')]
```

```python
colunas_linguagens
```

```python
popularidade_linguagens = df_final.groupby('ano')[colunas_linguagens].sum()
```

```python
popularidade_linguagens
```

```python
##porcentagem
total_respondentes_ano = df_final['ano'].value_counts()
popularidade_linguagensa_pct = popularidade_linguagens.div(total_respondentes_ano, axis=0).mul(100).round(2)
```

```python
popularidade_linguagensa_pct
```

```python
total_respondentes_ano
```

```python
popularidade_linguagensa_pct.to_csv(ANALISE_PARA_DASHBOARD/'analise_popularidade_linguagens_pct.csv')
print(f'Tabela gerada de % de uso da linguagem')
```
