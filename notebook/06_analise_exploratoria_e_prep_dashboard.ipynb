## 5 EXISTE DIFERENÇA SALARIAL POR GÊNERO (GENDER PAY GAP)?

```python
print('valores coluna gênero')
print(df_final['genero'].value_counts())
```

```python
analise_genero_senioridade = df_final.groupby(['ano', 'nivel_hierarquico', 'genero'])['salario_medio_mensal'].agg(['mean', 'median', 'count'])
```

```python
analise_genero_senioridade.columns = ['salario_medio', 'salario_mediana', 'contagem']
```

```python
analise_genero_senioridade[['salario_medio', 'salario_mediana']] = analise_genero_senioridade[['salario_medio', 'salario_mediana']].round(2)
```

```python
print("\nTabela gerada (Estatísticas Salariais por Ano e Gênero):")
print(analise_genero_senioridade.unstack(level='genero'))
```

```python
media_pivot_senioridade = analise_genero_senioridade['salario_medio'].unstack(level='genero')
```

```python
if 'Masculino' in media_pivot_senioridade.columns and 'Feminino' in media_pivot_senioridade.columns:
  media_pivot_senioridade['pay_gap_percentual'] = \
    ((media_pivot_senioridade['Masculino'] - media_pivot_senioridade['Feminino']) / media_pivot_senioridade['Masculino']) * 100
  
  pay_gap_por_nivel = media_pivot_senioridade[['Masculino', 'Feminino', 'pay_gap_percentual']].round(2)

  print("\nTabela Pay Gap por Nível (Diferença % entre Salário Médio Masculino e Feminino):")
  print(pay_gap_por_nivel)

else:
  print("\nNão foi possível calcular o Pay Gap por nível (colunas 'Masculino' ou 'Feminino' não encontradas).")
```

```python
analise_genero_senioridade.to_csv(ANALISE_PARA_DASHBOARD/'05_analise_salario_por_genero_senioridade_stats.csv')
print(f"\n>> Tabela detalhada por gênero e senioridade salva em {ANALISE_PARA_DASHBOARD}")
```

```python
if 'pay_gap_por_nivel' in locals(): 
  # Resetamos o índice para que 'ano' e 'nivel_hierarquico' virem colunas
  pay_gap_por_nivel.to_csv(ANALISE_PARA_DASHBOARD/'05_analise_pay_gap_percentual_por_nivel.csv')
  print(f">> Tabela do Pay Gap percentual por nível salva em {ANALISE_PARA_DASHBOARD}'/analise_pay_gap_percentual_por_nivel.csv'\n")
```
