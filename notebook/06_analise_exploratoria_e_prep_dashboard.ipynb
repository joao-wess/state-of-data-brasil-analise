## 9 - Qual a popularidade das plataformas Cloud por ano?

```python
print(df_final['clouds_utilizadas'].unique().tolist())
```

```python
df_final['clouds_utilizadas'].isna().sum()
```

```python
# ==============================================================================
# ANÁLISE 8 (CORRIGIDA): Qual a popularidade das plataformas Cloud por ano?
# ==============================================================================
print("\n--- Gerando Análise 8 (Corrigida): Popularidade das Plataformas Cloud ---")

# --- ETAPA 1: Verificar se 'clouds_utilizadas' existe ---
if 'clouds_utilizadas' not in df_final.columns:
    print("!! Erro: Coluna 'clouds_utilizadas' não encontrada no DataFrame final. Análise de Cloud não pode ser feita.")
else:
    # --- ETAPA 2: Limpeza Preliminar e Criação das Colunas Individuais ---
    # Criamos uma cópia temporária para segurança
    df_temp_cloud = df_final[['ano', 'clouds_utilizadas']].copy()
    df_temp_cloud['clouds_utilizadas_str'] = df_temp_cloud['clouds_utilizadas'].fillna('Nenhuma').astype(str).str.lower()

    plataformas = {
        'usa_cloud_aws': 'aws|amazon web services',
        'usa_cloud_gcp': 'gcp|google cloud',
        'usa_cloud_azure': 'azure', # Usando só 'azure' para evitar conflito com 'microsoft'
        'usa_cloud_oracle': 'oracle cloud',
        'usa_cloud_ibm': 'ibm', # Simplificado
        'usa_cloud_propria': 'cloud própria|propria',
        'usa_cloud_on_premise': 'on premise|on-premise|servidores on premise',
    }

    print("Criando colunas individuais para cada plataforma Cloud...")
    for nome_coluna, keyword in plataformas.items():
        df_temp_cloud[nome_coluna] = df_temp_cloud['clouds_utilizadas_str'].str.contains(keyword, case=False, regex=True).astype(int)

    # --- ETAPA 3: Tratamento da categoria 'Nenhuma' ---
    keywords_nenhuma = [
        'nenhuma', 'não utilizo', 'nao utilizo', 'não usamos', 'nao usamos', 
        'não uso', 'nao uso', 'nada', 'non applicable', 'n/a', 'nda'
    ]
    df_temp_cloud['usa_cloud_nenhuma_explicit'] = df_temp_cloud['clouds_utilizadas_str'].str.contains('|'.join(keywords_nenhuma), case=False, regex=True).astype(int)
    soma_clouds = df_temp_cloud[[col for col in plataformas.keys()]].sum(axis=1)
    df_temp_cloud['usa_cloud_nenhuma_implicita'] = (soma_clouds == 0).astype(int)
    df_temp_cloud['usa_cloud_nenhuma'] = ((df_temp_cloud['usa_cloud_nenhuma_explicit'] == 1) | (df_temp_cloud['usa_cloud_nenhuma_implicita'] == 1)).astype(int)

    # --- ETAPA 4: Calcular a popularidade ---
    colunas_cloud_final = list(plataformas.keys()) + ['usa_cloud_nenhuma']
    
    # Agrupamos usando o DataFrame temporário
    popularidade_cloud = df_temp_cloud.groupby('ano')[colunas_cloud_final].sum()
    total_respondentes_ano = df_temp_cloud['ano'].value_counts()
    popularidade_cloud_pct = popularidade_cloud.div(total_respondentes_ano, axis=0).mul(100).round(2)

    # --- ETAPA 5: AJUSTE PARA 2023 - MARCAR COMO NaN ---
    # Verificamos se a coluna original 'clouds_utilizadas' era majoritariamente nula em 2023
    if df_final[df_final['ano'] == 2023]['clouds_utilizadas'].isnull().mean() > 0.95: # Se > 95% for nulo
        print("Detectado que dados de uso de Cloud para 2023 não estão disponíveis no mesmo formato. Marcando como NaN.")
        # Define a linha de 2023 como NaN (Not a Number), exceto talvez 'NENHUMA' se fizer sentido
        popularidade_cloud_pct.loc[2023, popularidade_cloud_pct.columns != 'NENHUMA'] = np.nan 
        # Você pode decidir se quer manter o 100% em Nenhuma ou colocar NaN em tudo
        # popularidade_cloud_pct.loc[2023, :] = np.nan # Opção: NaN em tudo para 2023
        
    # --- ETAPA 6: Limpeza dos Nomes das Colunas ---
    nomes_limpos_cloud = {
        col: col.replace('usa_cloud_', '').replace('_', ' ').upper() 
        for col in popularidade_cloud_pct.columns
    }
    popularidade_cloud_pct.rename(columns=nomes_limpos_cloud, inplace=True)

    print("\nTabela gerada (% de usuários por plataforma Cloud - 2023 ajustado):")
    print(popularidade_cloud_pct)

    # --- PREPARAÇÃO PARA O DASHBOARD ---
    popularidade_cloud_pct.to_csv(ANALISE_PARA_DASHBOARD/'09_analise_popularidade_cloud_pct.csv')
```
